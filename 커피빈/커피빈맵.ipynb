```python
import requests

# 주소(문자열)를 입력받아 카카오 API를 통해 위치 정보를 반환하는 함수
def get_location(address):
    # 카카오 로컬 API 주소 (주소 검색용 endpoint)
    url = 'https://dapi.kakao.com/v2/local/search/address.json?query=' + address
    
    # API 인증키 (KakaoAK + 본인 REST API 키)
    headers = {
        'Authorization': 'KakaoAK dae86cabcb069c73392d01772a2ad095'
    }
    
    # requests.get()을 사용해 API 요청 -> 결과를 JSON으로 변환
    res = requests.get(url, headers=headers).json()

    # 좌표만 추출하고 싶다면:
    if res.get("documents"):
        x = res["documents"][0]["x"]  # 경도
        y = res["documents"][0]["y"]  # 위도
        print("위도:", y, "경도:", x)
    else: #검색이 실패했을 때 x = np.nan, y = np.nan으로 채우면 > 이후 데이터프레임으로 정리할 때 “빈값”으로 표시
        x = np.nan
        y = np.nan
    
    # 변환된 JSON 데이터를 반환
    return x, y
```

```python
x, y = get_location('서울시 강남구 영동대로 607')
x, y
```

```python
for i in df.index:
    address = df.loc[i, 'Address']
    x, y = get_location(address)
    df.loc[i, 'x'] = x
    df.loc[i, 'y'] = y
```

```python
import folium 

def get_map(address):
    filt = df['Address'].str.contains(address, na=False)  # 주소 검색
    store = df[filt]

    if len(store.index) == 0:  # 검색 결과 없을 때
        print('검색 주소 없음')
        return

    first = store.iloc[0]      # 첫 번째 검색 결과 선택
    lon = first['x']           # 경도
    lat = first['y']           # 위도

    map = folium.Map(location=[lat, lon], zoom_start=15)  # 지도 생성

    # ✅ 모든 검색 결과에 대해 마커 추가
    for _, row in store.iterrows():
        folium.Marker(
            location=[row['y'], row['x']], 
            icon=folium.Icon(color='red', icon='star')
        ).add_to(map)

    map.save(f'{address}.html')  # HTML로 저장

# 실행 예시
get_map('마포구')
```
